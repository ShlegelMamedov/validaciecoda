```python
#  Валидация пользовательского ввода
# Авторы: Шлегель Матвей, Мамедов Рамин | Группа 2-ИС33

import re

print("=" * 60)
print(" ПРАКТИЧЕСКАЯ РАБОТА №8: Валидация ввода")
print("=" * 60)

# =============================================================================
# ЗАДАНИЕ 1: Проверка ИНН (12 цифр для физлиц)
# =============================================================================
def validate_inn():
    """Запрашивает ИНН пока не будет введено 12 цифр."""
    while True:
        inn = input("\n Введите ИНН (12 цифр): ").strip()
        try:
            # Проверяем, что строка состоит только из цифр и имеет длину 12
            if re.fullmatch(r'\d{12}', inn):
                print(f" ИНН принят: {inn}")
                return inn
            else:
                print(" Ошибка: ИНН должен содержать ровно 12 цифр!")
        except Exception as e:
            print(f" Неожиданная ошибка: {e}")

# =============================================================================
# ЗАДАНИЕ 2: Валидация цены (положительное число ≤ 1 000 000)
# =============================================================================
def validate_price():
    """Запрашивает цену пока не будет введено корректное положительное число ≤ 1_000_000."""
    while True:
        price_input = input("\n Введите цену товара: ").strip()
        try:
            # Пробуем преобразовать в float (поддерживает целые и десятичные)
            price = float(price_input)

            # Проверка диапазона
            if price <= 0:
                print(" Ошибка: цена должна быть положительной!")
            elif price > 1_000_000:
                print(" Ошибка: цена не может превышать 1 000 000!")
            else:
                print(f" Цена принята: {price:,.2f} ₽")
                return price
        except ValueError:
            print(" Ошибка: введите корректное число (например, 150 или 99.99)!")
        except Exception as e:
            print(f" Неожиданная ошибка: {e}")

# =============================================================================
# ЗАДАНИЕ 3: Валидация пароля (8-20 символов, заглавная буква + цифра)
# =============================================================================
def validate_password():
    """Запрашивает пароль пока не будут соблюдены все требования."""
    while True:
        password = input("\n Введите пароль: ").strip()
        try:
            # Проверка длины
            if not (8 <= len(password) <= 20):
                print(" Ошибка: длина пароля должна быть от 8 до 20 символов!")
                continue

            # Проверка на заглавную букву
            if not re.search(r'[A-Z]', password):
                print(" Ошибка: пароль должен содержать хотя бы одну заглавную букву (A-Z)!")
                continue

            # Проверка на цифру
            if not re.search(r'\d', password):
                print(" Ошибка: пароль должен содержать хотя бы одну цифру (0-9)!")
                continue

            # Все проверки пройдены
            print(f" Пароль принят: {'*' * len(password)}")
            return password
        except Exception as e:
            print(f" Неожиданная ошибка: {e}")

# =============================================================================
# ТЕСТИРОВАНИЕ ВСЕХ ФУНКЦИЙ
# =============================================================================
print("\n ЗАПУСК ТЕСТИРОВАНИЯ")
print("-" * 60)

# Тест ИНН
print("\n[Тест 1/3] Валидация ИНН")
user_inn = validate_inn()

# Тест цены
print("\n[Тест 2/3] Валидация цены")
user_price = validate_price()

# Тест пароля
print("\n[Тест 3/3] Валидация пароля")
user_password = validate_password()

# =============================================================================
# ФИНАЛЬНАЯ СВОДКА
# =============================================================================
print("\n" + "=" * 60)
print(" РЕЗУЛЬТАТЫ ВАЛИДАЦИИ")
print("=" * 60)
print(f" ИНН: {user_inn}")
print(f" Цена: {user_price:,.2f} ₽")
print(f" Пароль: {'*' * len(user_password)} (длина: {len(user_password)})")
print("\n Все данные успешно проверены и приняты!")
```
